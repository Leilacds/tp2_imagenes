## Filosofía del MVP: Menos es Más

### ¿Qué es un MVP?

Un **Producto Mínimo Viable** (MVP) es la versión más simple de tu producto que resuelve el problema central de tu usuario. No es un prototipo ni una demo limitada, sino un producto **funcional** con las características esenciales.

### Principios del MVP

**1. Funcional sobre perfecto**
- Es mejor tener 3 funcionalidades que funcionan bien que 10 a medias
- Cada funcionalidad tiene que aportar valor real a la persona usuaria
- La simplicidad es una virtud, no una limitación

**2. Usuario primero**
- Diseñá para una persona específica, no para "usuarios en general"
- Entendé su problema antes de proponer la solución
- La documentación es para humanos, no solo para máquinas

**3. Iteración rápida**
- Empezá temprano, aprendé rápido, mejorá constantemente
- Es mejor tener algo que corre hoy que algo perfecto en un mes
- Cada versión tiene que poder ejecutarse de punta a punta

**4. Documentación = Producto**
- Un sistema sin documentación no está terminado
- Un README claro vale tanto como código limpio
- La experiencia de uso incluye mensajes de error, avisos y guías

## Más allá del Vibe Coding: Vibe Engineering

### Dos formas de trabajar con IA

Se espera, y se valora, que uses agentes de IA para desarrollar. Pero hay una diferencia enorme entre **usar IA** y **ser responsable de lo que la IA produce**.

- **Vibe coding**: describir lo que querés, aceptar lo que sale, pegar el error cuando falla y repetir. No hace falta leer el código. Sirve para prototipos descartables y para explorar ideas.
- **Vibe engineering**: término propuesto por **Simon Willison** (octubre de 2025) para el extremo profesional del espectro: usar agentes de código para ir más rápido, **sin renunciar a la responsabilidad por el resultado**. Quien lo practica define qué construir, divide el trabajo, exige pruebas, lee cada cambio y se hace cargo de lo que entrega.

Más tarde, Willison observó que el nombre que se está imponiendo en la industria es **agentic engineering**. Es la misma idea.

```
  VIBE CODING  ------------------------------------>  VIBE ENGINEERING
  ------------                                         ----------------
  Pido, acepto, pego, "funciona"                       Defino, delego, reviso, pruebo
  No leo lo que genera el agente                       Leo cada cambio (diff)
  Si falla, pido otro parche                           Si falla, entiendo por qué
  Sin pruebas, sin commits claros                      Pruebas mínimas, commits chicos
  Nadie responde por el resultado                      YO respondo por lo que entrego
```

**Referencia**: https://simonwillison.net/2025/Oct/7/vibe-engineering/

### Qué significa para este trabajo

El software que entregues lo firmás vos. En una defensa presencial con multiple choice no hay forma de improvisar lo que no entendiste. Por eso el trabajo se evalúa con una lógica de **vibe engineering**:

| Práctica | En tu proyecto |
|---|---|
| **Definir antes de delegar** | Escribí tu persona y tu declaración de proyecto *antes* de pedirle código a un agente |
| **Dividir en pasos chicos** | Un paso por vez: cargar imagen → procesar → comparar → analizar |
| **Pedir pruebas** | Una imagen de ejemplo y una verificación simple por cada función importante |
| **Leer cada cambio** | Revisá el diff antes de aceptarlo. Si no lo entendés, preguntale al agente o descartalo |
| **Commits chicos y descriptivos** | Un cambio, un commit. Tu historial cuenta tu proceso |
| **Documentar lo que decidiste** | En el README, anotá qué decidiste vos y qué sugirió el agente |
| **Dejar instrucciones al agente** | Un archivo de contexto del proyecto (`AGENTS.md` o `CLAUDE.md`) con tu stack, tus comandos y tus reglas |

## Real, Ficticio, Creíble, Verificable

Esta es la pregunta de fondo de este trabajo.

### Restaurar no es recuperar

Cuando una imagen pierde información (se achica, se pixela, se raya o pierde el color), esa información **no está más**. Un modelo de difusión no la "encuentra": la **genera de forma plausible**, a partir de lo que aprendió de millones de imágenes.

El resultado puede ser hermoso y convincente, y al mismo tiempo **inventado**.

```
   LA IMAGEN QUE VES  =  lo que había en tu original    (DATO)
                      +  lo que el modelo supone        (INVENTO)

   Lo difícil es que a simple vista no se distinguen.
```

### Cuatro palabras que no significan lo mismo

| Palabra | Pregunta que responde | Ejemplo |
|---|---|---|
| **Real** | ¿Estaba en la escena o en la imagen original? | Los píxeles que sobrevivieron al escaneo |
| **Ficticio** | ¿Lo agregó el modelo? | Las arrugas, las pestañas o el color de un vestido que nadie registró |
| **Creíble** | ¿Parece verdad? | Una cara nítida y coherente |
| **Verificable** | ¿Puedo comprobarlo con evidencia externa? | Comparar con otra foto, con un documento o con la persona fotografiada |

```
  ESCALERA DE LA CONFIANZA

  Verificable   <-- hay evidencia independiente que lo confirma
       ^
  Creíble       <-- parece natural, pero no hay forma de comprobarlo
       ^
  Plausible     <-- encaja con lo que el modelo aprendió
       ^
  Ficticio      <-- salió del modelo, no de la escena

  Una imagen restaurada casi nunca pasa de "creíble".
```

### Dos ejemplos para pensar

**1. Colorización.** Una foto en blanco y negro solo guarda la luminosidad de cada punto. El color del vestido de tu abuela **no está en la imagen**. Cualquier color que aparezca es una suposición del modelo. Es un caso donde la parte *real* (la luminosidad) y la parte *inventada* (el color) se pueden separar con claridad usando el espacio de color **L\*a\*b**: el canal L es dato y los canales a\*b\* son estimación.

**2. Super-resolución de una cara.** El modelo no "recupera" la cara original: dibuja una cara verosímil que, al reducirla, se parece a tu foto borrosa. Puede parecer más joven, más simétrica o... de otra persona.

```
    |                         |

                 |
        +--------+---------+
        |                  |
  QUÉ SE CONSERVÓ     QUÉ SE INVENTÓ
  (estructura,        (detalle fino,
   formas, bordes)     texturas, color)
                 |
        AVISO A LA PERSONA USUARIA
 "Esta imagen fue reconstruida por una IA.
  Revisá los detalles antes de usarla como evidencia."
```

- **Control de intensidad**: un deslizador de "cuánto puede cambiar el modelo mi imagen" (por ejemplo, la intensidad del procesamiento)
- **Comparador original/resultado** con deslizador
- **Mapa de diferencias**: restar el resultado del original (reescalado) y mostrar dónde cambió más
- **Aviso explícito** en la interfaz y en el archivo exportado ("restaurada con IA")
- **Línea base clásica**: mostrar también el resultado de un método tradicional (por ejemplo, interpolación bicúbica) para ver qué agrega el modelo
- **Descripción en lenguaje natural** de qué se conservó y qué se reconstruyó

No hace falta implementar todo. Elegí lo que le sirva a tu persona.

### Métricas: qué son y qué no son (explicación breve)

No son obligatorias, pero si querés ponerles números a las comparaciones, esto tenés que saber:

| Métrica | Qué mide, en simple | Necesita | Cuidado |
|---|---|---|---|
| **PSNR** | Qué tan parecidos son los píxeles de dos imágenes (a más alto, más parecidas) | Una imagen de **referencia** | Una imagen "linda" pero inventada puede tener PSNR bajo |
| **SSIM** | Qué tan parecida es la *estructura* (bordes, contraste, texturas) | Una imagen de **referencia** | Es mejor que PSNR, pero tampoco "ve" como una persona |
| **LPIPS** | Distancia de *percepción*, medida con una red neuronal (a más bajo, más parecidas) | Una imagen de **referencia** | Depende de la red usada |

**Idea clave**: las tres comparan contra una referencia. Con una foto real dañada **no tenés referencia**, así que no podés calcularlas.

Una forma honesta de usarlas: partí de una buena imagen, **degradala vos** (achicala, agregale ruido), restaurala y compará el resultado con la imagen original *sin degradar*. Así sabés cuánto se acercó el modelo a la verdad.

**Ninguna métrica te dice si algo es auténtico.** Mide parecido, no verdad.

## Track Único: Restauración y Mejora de Imágenes con Difusión

Todo el trabajo gira alrededor de **restaurar o mejorar imágenes** usando modelos de difusión disponibles en Hugging Face. Elegí **UNA** de estas tres sub-tareas:

---

### Sub-tarea A: Super-resolución

**Ideal si te interesa**: aumentar la resolución de imágenes pequeñas o de baja calidad.

**Casos de uso:** fotos escaneadas de baja resolución, capturas de pantalla, imágenes de archivo.

**Modelos de referencia:**
- `stabilityai/stable-diffusion-x4-upscaler` (difusión; sube la resolución 4 veces)
- Variantes con ControlNet para *upscaling* sobre Stable Diffusion / SDXL / FLUX (buscalas en el Hub y revisá la licencia)

**Conceptos de procesamiento digital:**
- Muestreo e interpolación (vecino más cercano, bilineal, bicúbica) como **línea base**
- Resolución espacial y aliasing
- Filtros de realce

---

### Sub-tarea B: Eliminación de ruido y desenfoque (denoising / deblurring)

**Ideal si te interesa**: limpiar imágenes con ruido, desenfocadas o con artefactos de compresión.

**Casos de uso:** fotos con poca luz, imágenes comprimidas, fotos movidas.

**Modelos de referencia:**
- Pipelines *image-to-image* de `diffusers` (por ejemplo, SDXL: `stabilityai/stable-diffusion-xl-base-1.0`) con **intensidad baja** (`strength` bajo) para que el modelo corrija sin reinventar
- ControlNet *tile* para conservar la estructura (por ejemplo, `lllyasviel/control_v11f1e_sd15_tile` sobre `stable-diffusion-v1-5/stable-diffusion-v1-5`)

**Conceptos de procesamiento digital:**
- Modelos de ruido (gaussiano, sal y pimienta)
- Filtros espaciales: media, mediana, gaussiano, bilateral
- Dominio de la frecuencia (Fourier) y filtros pasa-bajos / pasa-altos
- Restauración de imágenes

---

### Sub-tarea C: Fotos antiguas (daño y color)

**Ideal si te interesa**: restaurar fotos viejas con rayones, manchas, desvanecimiento o en blanco y negro.

**Casos de uso:** archivo familiar, fotos históricas, material de museos y bibliotecas.

**Modelos de referencia:**
- `black-forest-labs/FLUX.1-Kontext-dev` (edición guiada por instrucciones; requiere aceptar la licencia en Hugging Face y bastante memoria; hay versiones cuantizadas GGUF)
- `Qwen/Qwen-Image-Edit-2511` (edición por instrucciones; existe una versión acelerada `lightx2v/Qwen-Image-Edit-2511-Lightning`)

**Conceptos de procesamiento digital:**
- Histogramas y ecualización
- Espacios de color (RGB, L\*a\*b)
- *Inpainting* clásico con máscaras
- Segmentación y máscaras binarias

Esta sub-tarea es la que mejor muestra la pregunta central del trabajo: **el color de una foto en blanco y negro es 100% inventado**.

### Opción 1: Clasificación zero-shot (CLIP / SigLIP)

**Elegila si:** querés comparar la imagen con descripciones en lenguaje natural ("foto nítida", "foto con ruido") sin entrenar nada.

**Modelos:** `openai/clip-vit-base-patch32`, `google/siglip2-base-patch16-224`

```python
from transformers import pipeline

clasificador = pipeline(
    "zero-shot-image-classification",
    model="google/siglip2-base-patch16-224",
)

categorias = [
    "una foto nítida y con detalle",
    "una foto borrosa",
    "una foto con mucho ruido",
    "una foto pixelada",
]

resultado = clasificador(imagen_procesada, candidate_labels=categorias)
```

**Cuidado:** estos modelos no fueron entrenados para medir calidad. Las frases influyen mucho en el resultado y las diferencias entre "antes" y "después" pueden ser pequeñas. Usalos para **comparar** original y resultado, no como un juez absoluto.

---

### Opción 2: Detección de objetos (DETR / RT-DETR)

**Elegila si:** querés verificar que el contenido **se preservó** (que no aparecieron ni desaparecieron objetos o personas).

**Modelos:** `facebook/detr-resnet-50`, `PekingU/rtdetr_r50vd`

```python
from transformers import pipeline

detector = pipeline("object-detection", model="PekingU/rtdetr_r50vd")

antes = detector(imagen_original)
despues = detector(imagen_procesada)

# Comparar qué objetos se detectan antes y después
```

**Para qué sirve en tu proyecto:** si el original tenía 2 personas y el resultado tiene 3, hay algo para explicar. Esto es verificación en acción.

---

### Opción 3: Descripción con un modelo de visión y lenguaje (VLM)

**Elegila si:** querés una explicación en palabras de qué cambió.

**Modelos chicos, para correr en local:** `HuggingFaceTB/SmolVLM2-256M-Video-Instruct`, `HuggingFaceTB/SmolVLM2-2.2B-Instruct`, `Qwen/Qwen2.5-VL-3B-Instruct`

```python
from transformers import pipeline

descriptor = pipeline(
    "image-text-to-text",
    model="HuggingFaceTB/SmolVLM2-2.2B-Instruct",
)

mensajes = [{
    "role": "user",
    "content": [
        {"type": "image", "image": imagen_procesada},
        {"type": "text", "text": "Describí esta imagen en una oración. ¿Se ven rostros, texto o paisajes?"},
    ],
}]

respuesta = descriptor(text=mensajes, max_new_tokens=100)
```

**Cuidado:** los modelos chicos se equivocan y "alucinan". Tratalos como una ayuda para la persona usuaria, **nunca como prueba**. Es otro ejemplo de la tensión entre *creíble* y *verificable*.

(Consultá la documentación actual de `transformers` para el formato exacto de la entrada de cada modelo; cambia entre versiones.)

### Optimizaciones de Performance

**1. Cargá el modelo una sola vez**

En Gradio, cargalo al nivel del módulo (fuera de la función que se ejecuta con cada clic):
```python
pipe = cargar_modelo()   # una vez, al iniciar

def restaurar(imagen, intensidad):
    return pipe(imagen=imagen, strength=intensidad).images[0]
```
En Streamlit, usá `@st.cache_resource`.

**2. Mostrá progreso**
```python
import gradio as gr

def procesar(imagen, progreso=gr.Progress()):
    progreso(0.1, desc="Preparando la imagen...")
    # ...
    progreso(0.6, desc="Reconstruyendo detalles con el modelo de difusión...")
    # ...
    return resultado
```

**3. Evitá reprocesar**

Guardá el resultado en el estado de la sesión (`gr.State` en Gradio, `st.session_state` en Streamlit) para no volver a procesar cada vez que cambian un control de visualización.

---

### Depuración Efectiva

**Técnica 1: Mensajes de error comprensibles**
```python
try:
    resultado = restaurar(imagen, intensidad)
except Exception as error:
    raise gr.Error("No pude procesar la imagen. Probá con una más chica o en formato JPG.") from error
```

**Técnica 2: Registros (logs) en la consola**
```python
import logging

logging.basicConfig(level=logging.INFO)
registro = logging.getLogger(__name__)
registro.info("Iniciando procesamiento (%sx%s)", ancho, alto)
```
